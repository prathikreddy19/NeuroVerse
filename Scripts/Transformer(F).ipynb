```python
import torch
import json
import os

# Define the path
path = '/content/regression_models/regression_models'

# Load config
with open(os.path.join(path, 'config.json'), 'r') as f:
    config = json.load(f)

# Simple regression model class (adjust based on your architecture)
class RegressionModel(torch.nn.Module):
    def __init__(self, input_size):
        super(RegressionModel, self).__init__()
        self.linear = torch.nn.Linear(input_size, 1)
    def forward(self, x):
        return self.linear(x)

# List of model names
model_names = ["EPS", "EEA", "EGB", "EI", "NC", "XC"]
models = {}

# Load each model
for name in model_names:
    model = RegressionModel(config['input_size'])  # Adjust based on config
    model.load_state_dict(torch.load(os.path.join(path, f'{name}_weights.pt')))
    model.eval()
    models[name] = model

# Example test input (replace with your actual test data)
test_input = torch.randn(10, config['input_size'])
for name, model in models.items():
    with torch.no_grad():
        output = model(test_input)
        print(f"{name} output shape: {output.shape}, sample output: {output[0]}")
```
